```python
import pandas as pd

print("Pandas version:", pd.__version__)
```

```python
# 2. Load Dataset
# Purpose: Load the CSV file into a Pandas DataFrame.

df = pd.read_csv("../data/pricing_discount_data.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

```python
# 3. Inspect Dataset Structure
# Purpose: Check column names and data types.

df.info()
```

```python
# 4. Preview Dataset
# Purpose: View the first five records.

df.head()
```

```python
# 5. View Last Five Rows
# Purpose: Preview the final five records.

df.tail()
```

```python
# 6. Check Column Names
# Purpose: Display all column names in the dataset.

df.columns
```

```python
# 7. Summary Statistics
# Purpose: Understand the distribution of numerical columns.

df.describe()
```

```python
# 8. Check Duplicate Rows
# Purpose: Count completely duplicated rows.

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# 9. Check Missing Values
# Purpose: Count missing values in each column.

df.isnull().sum()
```

```python
# 10. Check Categorical Values
# Purpose: Identify unique categories, regions, and sales channels.

print("Categories:", df["category"].unique())
print("Regions:", df["region"].unique())
print("Channels:", df["channel"].unique())
```

```python
# 11. Check Unique Order IDs
# Purpose: Check whether order IDs are repeated.

print("Unique order IDs:", df["order_id"].nunique())
print("Total rows:", len(df))
```

```python
# 12. Check Date Range
# Purpose: Find the earliest and latest order dates.

df["order_date"] = pd.to_datetime(df["order_date"])

print("Earliest order date:", df["order_date"].min())
print("Latest order date:", df["order_date"].max())
```

```python
# 13. Validate Discount Percentages
# Purpose: Check for discount values outside the 0–100% range.

print("Minimum discount:", df["discount_pct"].min())
print("Maximum discount:", df["discount_pct"].max())

print("Invalid discount rows:",
      ((df["discount_pct"] < 0) | (df["discount_pct"] > 1)).sum())
```

```python
# 14. Validate Quantity
# Purpose: Check for zero or negative quantities.

print("Minimum quantity:", df["quantity"].min())
print("Invalid quantity rows:", (df["quantity"] <= 0).sum())
```

```python
# 15. Validate Prices and Costs
# Purpose: Count negative values in key price and cost columns.

for column in ["list_price", "unit_cost", "selling_price"]:
    print(column, "negative values:", (df[column] < 0).sum())
```

```python
# 16. Check Zero Prices and Costs
# Purpose: Count zero values in key price and cost columns.


for column in ["list_price", "unit_cost", "selling_price"]:
    print(column, "zero values:", (df[column] == 0).sum())
```

```python
# 17. Validate Selling Price Calculation

expected_selling_price = (
    df["list_price"] * (1 - df["discount_pct"])
).round(2)

mismatch_count = (
    expected_selling_price != df["selling_price"].round(2)
).sum()

print("Selling price mismatches:", mismatch_count)
```

```python
# 18. Validate Gross Revenue Calculation

expected_gross_revenue = (
    df["net_revenue"] + df["discount_amount"]
).round(2)

mismatch_count = (
    expected_gross_revenue != df["gross_revenue"].round(2)
).sum()

print("Gross revenue mismatches:", mismatch_count)
```

```python
# Validate Total Cost Calculation

expected_total_cost = (
    df["unit_cost"] * df["quantity"]
).round(2)

mismatch_count = (
    expected_total_cost != df["total_cost"].round(2)
).sum()

print("Total cost mismatches:", mismatch_count)
```

```python
# Validate Profit Calculation

expected_profit = (
    df["net_revenue"] - df["total_cost"]
).round(2)

mismatch_count = (
    expected_profit != df["profit"].round(2)
).sum()

print("Profit mismatches:", mismatch_count)
```

```python
# 19. Check Negative Profit

negative_profit_count = (df["profit"] < 0).sum()

print("Loss-making orders:", negative_profit_count)
```

```python
# 20. Count Profitable Orders

profitable_count = (df["profit"] > 0).sum()

print("Profitable orders:", profitable_count)
```
